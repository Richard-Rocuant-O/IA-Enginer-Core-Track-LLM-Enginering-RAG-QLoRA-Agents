# Explicando el 'kernel'

Detrás de este notebook hay un proceso de Python que ejecuta cada celda cuando la corres. Ese proceso de Python se conoce como el Kernel. Cada notebook tiene su propio Kernel independiente.

Puedes hacer clic en el botón de arriba "Restart Kernel" (Reiniciar Kernel).

Si luego intentas ejecutar la siguiente celda, obtendrás un error, porque favorite_fruit ya no está definida. Tendrás que ejecutar las celdas desde el principio del notebook otra vez. Entonces la siguiente celda debería ejecutarse sin problema.

````python
print(f"My favorite fruit is {favorite_fruit}")
````

# Salida de la celda (cell output)

Cuando ejecutas una celda, la salida estándar y el resultado de la última instrucción se escriben en el área inmediatamente debajo del código, conocida como la 'salida de la celda' (cell output). Cuando guardas un Notebook desde el menú de archivo (o ctrl+S o command+S), la salida también se guarda, lo que lo convierte en un registro útil de lo que sucedió.

Puedes limpiar esto haciendo clic en "Clear All Outputs" (Limpiar todas las salidas) en la barra de herramientas. Es buena idea limpiar las salidas antes de subir el código a un repositorio como GitHub, de lo contrario los archivos pueden volverse grandes y más difíciles de leer.

````python
spams = ["spam"] * 1000
print(spams)

# ¡Podría valer la pena limpiar la salida después de ejecutar esto!
````

# Pequeños detalles que encontramos en el curso

Esto no es necesariamente una función de los notebooks, pero es un paquete agradable de conocer que es útil en los notebooks.

El paquete `tqdm` imprimirá una bonita barra de progreso si envuelves cualquier iterable con él.

````python
# Aquí hay código sin barra de progreso
# Tardará 10 segundos mientras te preguntas qué está pasando..

import time

spams = ["spam"] * 1000

for spam in spams:
    time.sleep(0.01)
````

````python
# Y ahora, con una bonita barra de progreso:

import time
from tqdm import tqdm

spams = ["spam"] * 1000

for spam in tqdm(spams):
    time.sleep(0.01)
````

````python
# Sobre un tema diferente, aquí hay una forma útil de imprimir salida en markdown

from IPython.display import Markdown, display

display(Markdown("# ¡Este es un encabezado grande!\n\n- Y este es un punto de viñeta\n- Este también\n- ¡Y yo también!"))
````
