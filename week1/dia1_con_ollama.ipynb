## Bien, sigamos adelante con nuestro primer proyecto

````python
# Una clase para representar una página web
# Si no estás familiarizado con las Clases, revisa el notebook "Intermediate Python"

# Algunos sitios web requieren que uses headers apropiados al obtenerlos:
headers = {
 "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36"
}

class Website:

    def __init__(self, url):
        """
        Crea este objeto Website a partir de la url dada usando la librería BeautifulSoup
        """
        self.url = url
        response = requests.get(url, headers=headers)
        soup = BeautifulSoup(response.content, 'html.parser')
        self.title = soup.title.string if soup.title else "No title found"
        for irrelevant in soup.body(["script", "style", "img", "input"]):
            irrelevant.decompose()
        self.text = soup.body.get_text(separator="\n", strip=True)
````

````python
# Probemos con uno. Cambia el sitio web y agrega instrucciones print para seguir el proceso.

ed = Website("https://edwarddonner.com")
print(ed.title)
print(ed.text)
````

## Tipos de prompts

Puede que ya sepas esto - pero si no, ¡te familiarizarás mucho con ello!

Modelos como GPT4o han sido entrenados para recibir instrucciones de una forma particular.

Esperan recibir:

**Un system prompt** que les dice qué tarea están realizando y qué tono deberían usar

**Un user prompt** -- el inicio de la conversación al que deben responder

````python
# Define nuestro system prompt - puedes experimentar con esto después, cambiando la última oración a 'Responde en markdown en español."

system_prompt = "Eres un asistente que analiza el contenido de un sitio web \
y ofrece un resumen corto, ignorando el texto que pueda ser de navegación. \
Responde en markdown."
````

````python
# Una función que escribe un User Prompt que pide resúmenes de sitios web:

def user_prompt_for(website):
    user_prompt = f"Estás viendo un sitio web titulado {website.title}"
    user_prompt += "\nEl contenido de este sitio web es el siguiente; \
por favor proporciona un resumen corto de este sitio web en markdown. \
Si incluye noticias o anuncios, resúmelos también.\n\n"
    user_prompt += website.text
    return user_prompt
````

````python
print(user_prompt_for(ed))
````

## Mensajes

La API de OpenAI espera recibir mensajes en una estructura particular.
Muchas de las otras APIs comparten esta estructura:

```
[
    {"role": "system", "content": "aquí va el mensaje del sistema"},
    {"role": "user", "content": "aquí va el mensaje del usuario"}
]

Para darte un adelanto, las siguientes 2 celdas hacen una llamada bastante simple - ¡no exigiremos al todopoderoso GPT (todavía)!

````python
messages = [
    {"role": "system", "content": "Eres un asistente sarcástico"},
    {"role": "user", "content": "¿Cuánto es 2 + 2?"}
]
````

````python
# Para darte un adelanto -- llamar a OpenAI con mensajes de system y user:

response = openai.chat.completions.create(model="llama3.2", messages=messages)
print(response.choices[0].message.content)
````

## Y ahora construyamos mensajes útiles para GPT-4o-mini, usando una función

````python
# Mira cómo esta función crea exactamente el formato de arriba

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_for(website)}
    ]
````

````python
# Prueba esto, y luego intenta con algunos otros sitios web

messages_for(ed)
````

## Hora de juntar todo - ¡la API de OpenAI es muy simple!

````python
# Y ahora: llama a la API de OpenAI. ¡Te familiarizarás mucho con esto!

def summarize(url):
    website = Website(url)
    response = openai.chat.completions.create(
        model = "llama3.2",
        messages = messages_for(website)
    )
    return response.choices[0].message.content
````

````python
summarize("https://edwarddonner.com")
````

````python
# Una función para mostrar esto de forma más agradable en la salida de Jupyter, usando markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
````

````python
display_summary("https://edwarddonner.com")
````
